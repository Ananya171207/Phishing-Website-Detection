```python
from sklearn.model_selection import train_test_split

# Drop 'url' column only if it exists, to prevent KeyError on re-runs
if "url" in df.columns:
    df = df.drop("url",axis=1)

x=df.drop("status",axis=1)
y=df["status"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape,y_train.shape)
print(x_test.shape,y_test.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, mean_squared_error, mean_absolute_error, r2_score

```

```python
from sklearn.linear_model import LogisticRegression
import pandas as pd
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

logreg = LogisticRegression(max_iter=5000)
logreg.fit(x_train_scaled, y_train)
y_pred_lr = logreg.predict(x_test_scaled)


y_pred_lr_enc = pd.Series(y_pred_lr).map({'legitimate': 0, 'phishing': 1})
print("metric of logistic regression")
print("Accuracy: ", accuracy_score(y_test, y_pred_lr))
print("Precision: ", precision_score(y_test, y_pred_lr, pos_label='phishing'))
print("Recall: ", recall_score(y_test, y_pred_lr, pos_label='phishing'))
print("F1 score: ", f1_score(y_test, y_pred_lr, pos_label='phishing'))
print("rsmne:", np.sqrt(mean_squared_error(y_test_enc, y_pred_lr_enc)))
print("mae:", mean_absolute_error(y_test_enc, y_pred_lr_enc))
print("r2 score:", r2_score(y_test_enc, y_pred_lr_enc))
```

```python
dt = DecisionTreeClassifier()
dt.fit(x_train, y_train)
y_pred_dt = dt.predict(x_test)


y_pred_dt_enc = pd.Series(y_pred_dt).map({'legitimate': 0, 'phishing': 1})
print("metric for decision tree")
print("Accuracy: ", accuracy_score(y_test, y_pred_dt))
print("Precision: ", precision_score(y_test, y_pred_dt, pos_label='phishing'))
print("Recall: ", recall_score(y_test, y_pred_dt, pos_label='phishing'))
print("F1 score: ", f1_score(y_test, y_pred_dt, pos_label='phishing'))
print("rsmne:", np.sqrt(mean_squared_error(y_test_enc, y_pred_dt_enc)))
print("mae:", mean_absolute_error(y_test_enc, y_pred_dt_enc))
print("r2 score:", r2_score(y_test_enc, y_pred_dt_enc))
```

```python
rf = RandomForestClassifier()
rf.fit(x_train, y_train)
y_pred_rf = rf.predict(x_test)


y_pred_rf_enc = pd.Series(y_pred_rf).map({'legitimate': 0, 'phishing': 1})
print("metric for Random Forest")
print("Accuracy: ", accuracy_score(y_test, y_pred_rf))
print("Precision: ", precision_score(y_test, y_pred_rf, pos_label='phishing'))
print("Recall: ", recall_score(y_test, y_pred_rf, pos_label='phishing'))
print("F1 score: ", f1_score(y_test, y_pred_rf, pos_label='phishing'))
print("rsmne:", np.sqrt(mean_squared_error(y_test_enc, y_pred_rf_enc)))
print("mae:", mean_absolute_error(y_test_enc, y_pred_rf_enc))
print("r2 score:", r2_score(y_test_enc, y_pred_rf_enc))
```

```python
from lightgbm import LGBMClassifier

import numpy as np


y_train_enc = y_train.map({'legitimate': 0, 'phishing': 1})
y_test_enc = y_test.map({'legitimate': 0, 'phishing': 1})


lgb_model = LGBMClassifier(random_state=42, verbose=-1)
lgb_model.fit(x_train, y_train_enc)


lgb_preds = lgb_model.predict(x_test)
print("metrics for LightGBM")
print(f"Accuracy:  {accuracy_score(y_test_enc, lgb_preds):.5f}")
print(f"Precision: {precision_score(y_test_enc, lgb_preds):.5f}")
print(f"Recall:    {recall_score(y_test_enc, lgb_preds):.5f}")
print(f"F1 Score:  {f1_score(y_test_enc, lgb_preds):.5f}")
print(f"RMSE:      {np.sqrt(mean_squared_error(y_test_enc, lgb_preds)):.5f}")
print(f"MAE:       {mean_absolute_error(y_test_enc, lgb_preds):.5f}")
print(f"R2 Score:  {r2_score(y_test_enc, lgb_preds):.5f}")
```

```python
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, mean_squared_error, mean_absolute_error, r2_score
import numpy as np


y_train_enc = y_train.map({'legitimate': 0, 'phishing': 1})
y_test_enc = y_test.map({'legitimate': 0, 'phishing': 1})


cat_model = CatBoostClassifier(random_state=42, verbose=0)
cat_model.fit(x_train, y_train_enc)


cat_preds = cat_model.predict(x_test)


print(" metric for CatBoost ")
print(f"Accuracy:  {accuracy_score(y_test_enc, cat_preds):.5f}")
print(f"Precision: {precision_score(y_test_enc, cat_preds):.5f}")
print(f"Recall:    {recall_score(y_test_enc, cat_preds):.5f}")
print(f"F1 Score:  {f1_score(y_test_enc, cat_preds):.5f}")
print(f"RMSE:      {np.sqrt(mean_squared_error(y_test_enc, cat_preds)):.5f}")
print(f"MAE:       {mean_absolute_error(y_test_enc, cat_preds):.5f}")
print(f"R2 Score:  {r2_score(y_test_enc, cat_preds):.5f}")
```
